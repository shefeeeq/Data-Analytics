```python
#create a tuple containing the following numbers. cound the even and odd numbers seperatly and their sum.
numbers=(12,25,36,41,58,63,70,89)
even_num=0
even_sum=0
odd_num=0
odd_sum=0
for num in numbers:
    if num %2==0:
        even_num += 1
        even_sum+=num
    else:
        odd_num+=1
        odd_sum+=num
print("Even numbers :",even_num)
print("Even numbers sum :",even_sum)
print("Odd numbers :",odd_num)
print("Odd numbers sum : ",odd_sum)
```
