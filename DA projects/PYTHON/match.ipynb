```python
#display the season based on a month number
month=int(input("enter the month number"))

match month:
    case 3 | 4 | 5:
        print("Season: Spring")
    case 6 | 7 | 8:
        print("Season: Summer")
    case 9 | 10 | 11:
        print("Season: Autumn")
    case 12 | 1 | 2:
        print("Season: Winter")
    case _:
        print("Invalid month")
```

```python
# create a driven prigram to select drink (tea , coffe, water , juice using match.
print("Drink Menu:")
print("1. Tea")
print("2. Coffee")
print("3. Water")
print("4. Juice")

choice = int(input("Enter your choice (1-4): "))

match choice:
    case 1:
        print("You selected Tea")
    case 2:
        print("You selected Coffee")
    case 3:
        print("You selected Water")
    case 4:
        print("You selected Juice")
    case _:
        print("Invalid choice")
```

```python
#display the result of a students attendans statu based on the input (P->presant, A->absent ,L->leave)
status = input("Enter attendance status (P/A/L): ")

# Match-case structure
match status:
    case "P":
        print("Student is Present")
    case "A":
        print("Student is Absent")
    case "L":
        print("Student is on Leave")
    case _:
        print("Invalid input! Please enter P, A, or L.")
```
